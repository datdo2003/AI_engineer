```python
from transformers import pipeline
```

```python
sentiment_classifier = pipeline("sentiment-analysis")
sentiment_classifier("I'm very happy to work with AI")
```

```python
sentiment_classifier("the weather today is cloudy")
```

```python
sentiment_classifier("the door is open")
```

```python
ner = pipeline("ner", model="dslim/bert-base-NER")
ner("I work in Ho Chi Minh City and I love my job")
```

```python
ner("Her name is Anna and she works in New York City for Morgan Stanley")
```

```python
from transformers import AutoTokenizer
model = "bert-base-uncased"
tokenizer = AutoTokenizer.from_pretrained(model)
```

```python
sentence = "I'm so excited to be learning about large language models"
input_ids = tokenizer(sentence)
print(input_ids)
```

```python
print (tokenizer.decode(101))
print (tokenizer.decode(1045))
```

```python
tokens = tokenizer.tokenize(sentence)
print (tokens)
```

```python
token_ids = tokenizer.convert_tokens_to_ids(tokens)
print (token_ids)
```
