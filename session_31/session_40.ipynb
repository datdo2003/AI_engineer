# Learning about output parser

```python
%load_ext dotenv
%dotenv
```

```python
from langchain_openai import ChatOpenAI
from langchain_core.output_parsers import StrOutputParser, CommaSeparatedListOutputParser

chat = ChatOpenAI(model="gpt-4",
                  temperature=0,
                  max_completion_tokens=500,
                  seed=365)
response = chat.invoke('''Hãy cho tôi biết điều gì đó về đại dương mà đa số mọi người không biêt''')
```

```python
# response
str_output_parser = StrOutputParser()
str_output_parser.invoke(response)
```

### Comma Seperated output parser

```python
from langchain_core.output_parsers import CommaSeparatedListOutputParser
comma_output_parser = CommaSeparatedListOutputParser()
comma_output_parser.invoke(response)
```

### Date time output parser

```python
comma_output_parser.get_format_instructions()
```

```python
response = chat.invoke(f'''Tôi vừa nhận nuôi một con chó, bạn có thể gợi ý cho tôi vài cái tên hài hước được không?
{CommaSeparatedListOutputParser().get_format_instructions()}''')
```

```python
comma_output_parser.invoke(response)
```
