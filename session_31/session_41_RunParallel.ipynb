# RunParallel

```python
%load_ext dotenv
%dotenv
```

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_openai import ChatOpenAI
from langchain_core.output_parsers import StrOutputParser
from langchain_core.runnables import RunnableParallel
```

```python
chat = ChatOpenAI(model='gpt-4',
                  seed=365,
                  max_completion_tokens=500,
                  temperature=0)
```

```python
prompt_template_book = ChatPromptTemplate.from_template('''Cho tôi 5 cuốn sách hay nhất về ngôn ngữ lập trình {programming language}.
Chỉ liệt kê tên.''')
prompt_template_project = ChatPromptTemplate.from_template('''Cho tôi 5 dự án hay nhất về ngôn ngữ {programming language}. 
Chỉ liệt kê danh sách.''')

prompt_template_time = ChatPromptTemplate.from_template('''Tôi là một developer có level intermediate.
Xem xét các cuốn sách
{books}

Và xem xet các projects
{projects}

Hãy cho tôi biết một cách sơ bộ thời gian cần thiết để tôi có thể hiểu chúng.
''')
```

```python
str_output_parser = StrOutputParser()
```

```python
book_chain = prompt_template_book | chat | str_output_parser
project_chain = prompt_template_project | chat | str_output_parser
# book_chain.invoke({'programming language': 'python'})
```

```python
parallel = RunnableParallel({'books': book_chain, 'projects': project_chain})
parallel.invoke({'programming language': 'python'})
```

```python
parallel.get_graph().print_ascii()
```

```python
chain_time = parallel | prompt_template_time | chat | str_output_parser
print(chain_time.invoke({'programming language': 'python'}))
```

```python
chain_time.get_graph().print_ascii()
```

```python
chain_time1 = {'books': book_chain, 'projects': project_chain} | prompt_template_time | chat | str_output_parser
```

```python
print(chain_time1.invoke({'programming language': 'python'}))
```
