```python
import os

%load_ext dotenv
%dotenv
```

```python
from langchain_openai import ChatOpenAI
from langchain_core.prompts import SystemMessagePromptTemplate, HumanMessagePromptTemplate, ChatPromptTemplate
chat = ChatOpenAI(model="gpt-4",
                  max_completion_tokens=100,
                  seed=365,
                  temperature=0)

TEMPLATE_S = '''{description}'''
TEMPLATE_H = '''Tôi sắp nhận nôi một con {pet}, bạn hãy gợi ý cho con {pet} vài cái tên thật đáng yêu.'''
system_message_template = SystemMessagePromptTemplate.from_template(TEMPLATE_S)
human_message_template = HumanMessagePromptTemplate.from_template(TEMPLATE_H)
# system_message_template
# human_message_template
chat_prompt_template = ChatPromptTemplate.from_messages([system_message_template, human_message_template])
chat_prompt_value = chat_prompt_template.invoke({"description": "Bạn là một chuyên gia đặt tên cho động vật", "pet": "chó"})
# chat_prompt_value
```

```python
response = chat.invoke(chat_prompt_value)
response.content
```
